```python
import json
try:
    with open("library.json", "r") as file:
        library = json.load(file)
except FileNotFoundError:
    library = []

print(library)
```

```python
def save_records():
    with open("library.json", "w") as file:
        json.dump(library, file, indent=4)
```

```python
save_records()
```

```python
def add_book():
    book_id = int(input("Enter book id: "))
    title = input("Enter title of the book :")
    author = input("Enter author's name: ")

    book = {
        "id" : book_id ,
        "title" : title ,
        "autthor" : author , 
        "available" : True 
    
    }
    library.append(book)
    save_records()

    print("Book added successfully ! ")
```

```python
def display_available_books():
    print("\n Available books")
    print("-" * 60)
    for book in library:
        if book["available"]:
            print(f"ID: {book['id']}")
            print(f"Title: {book['title']}")
            print(f"Author: {book['author']}")
            print("-" * 60)
```

```python
def search_book(): 
    search_title = input("Enter name of the book you want to search :")
    found = False

    for book in library:
        if search_title.lower() in book["title"].lower():
            print( "Book found !")
            print(f"ID: {book['id']}")
            print(f"Title: {book['title']}")
            print(f"Author: {book['author']}")
            
            if book["available"]:
                print("Status: Available")
            else:
                print("Status: Issued")

            found = True

    if not found:
        print("Book not found.")
```

```python
def issue_book():
    book_id = int(input("Enter book ID to issue: "))

    for book in library:
        if book["id"] == book_id: 
            if book["available"]: 
                book["available"] = False
                save_records()
                print ("Book issued successfully!")
            else:
                print ("Book is already issued!" )
                      
            return

    print("Book not found!")
    
```

```python
def return_book(): 
    book_id = int(input("Enter book id to return:"))

    for book in library: 
        if book["id"] == book_id: 
            if not book ["available"]: 
                book["available"] = True
                save_records()
                print("Book returned successfully!")
            else: 
                print("Book is already available!")

            return

    print("Book not found!")
    
```

```python
def library_management_system():
    while True:
        print("\n===== LIBRARY MANAGEMENT SYSTEM =====")
        print("1. Add Book")
        print("2. Search Book")
        print("3. Issue Book")
        print("4. Return Book")
        print("5. Display Available Books")
        print("6. Save Records")
        print("7. Exit")

        choice = input("Enter your choice: ")

        if choice == "1":
            add_book()

        elif choice == "2":
            search_book()

        elif choice == "3":
            issue_book()

        elif choice == "4":
            return_book()

        elif choice == "5":
            display_available_books()

        elif choice == "6":
            save_records()
            print("Records saved successfully!")

        elif choice == "7":
            save_records()
            print("Thank you for using the Library Management System!")
            break

        else:
            print("Invalid choice. Please try again.")
```
